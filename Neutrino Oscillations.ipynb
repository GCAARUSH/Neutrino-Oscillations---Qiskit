# Neutrino Oscillations

## 1. What is a neutrino?

A neutrino is a **fundamental particle** (as far as we know it is not made of anything smaller) belonging to the family of **leptons**, together with the electron, the muon and the tau.

Its main properties:

- It is a **spin-½ fermion**.
- It has **no electric charge**.
- It feels only the **weak nuclear force** (and gravity), so it passes through ordinary matter almost untouched. Roughly 60 billion solar neutrinos cross every square centimeter of your body every second, and nearly all of them go straight through.
- It has an **extremely small, but non-zero, mass**, at least a million times lighter than the electron.

### The three flavors

| Generation | Charged lepton | Neutrino (its "flavor") |
|:---:|:---:|:---:|
| 1 | electron $e^-$ | electron neutrino $\nu_e$ |
| 2 | muon $\mu^-$ | muon neutrino $\nu_\mu$ |
| 3 | tau $\tau^-$ | tau neutrino $\nu_\tau$ |

The **flavor** of a neutrino is the charged lepton it is partnered with in a weak interaction. For example, a $\nu_\mu$ that hits a nucleus can produce a muon, but not an electron. Flavor is therefore how we *label* a neutrino at the moment it is **created** and at the moment it is **detected**.

### Where do neutrinos come from?

| Source | Typical flavor | Typical energy |
|:---|:---:|:---:|
| The Sun (nuclear fusion) | $\nu_e$ | 0.1 to 10 MeV |
| Nuclear reactors | $\bar{\nu}_e$ | a few MeV |
| Atmosphere (cosmic-ray showers) | $\nu_\mu$ and $\nu_e$ | 0.1 to 100 GeV |
| Accelerator beams | $\nu_\mu$ | 1 to a few GeV |
| Supernovae | all flavors | about 10 MeV |

## 2. The problem that led to the discovery of flavors
The solar neutrino problem arose when experiments detected significantly **fewer electron neutrinos** from the Sun than predicted by theoretical models of solar fusion. Since the Sun produces electron neutrinos, scientists initially suspected either errors in solar models or problems with neutrino detection. The mystery was resolved through the discovery of **neutrino oscillations**: as neutrinos travel from the Sun to Earth, they can **change flavour** from electron neutrinos into muon or tau neutrinos. Early experiments primarily detected electron neutrinos, missing the others. This discovery demonstrated that neutrinos have **non-zero masses** and provided important evidence for physics beyond the original Standard Model.

## 3. The key idea: flavor states vs. mass states

A neutrino can be described in two different "languages" (bases):

| | **Flavor states** $\nu_e, \nu_\mu, \nu_\tau$ | **Mass states** $\nu_1, \nu_2, \nu_3$ |
|:---|:---|:---|
| What defines them | Which charged lepton they pair with in weak interactions | They have a definite mass $m_1, m_2, m_3$ |
| When they matter | Creation and detection | Travel through space |

These two sets are **not the same**. Each flavor state is a *superposition* of the three mass states:

$$|\nu_\alpha\rangle = \sum_{i=1}^{3} U_{\alpha i}^{*}\,|\nu_i\rangle ,\qquad \alpha = e,\mu,\tau$$

where $U$ is a $3\times 3$ unitary matrix called the **PMNS mixing matrix** (Pontecorvo–Maki–Nakagawa–Sakata).

A mass state has a definite energy for a given momentum, $E_i=\sqrt{p^2+m_i^2}$, so as time passes each one picks up its own quantum phase $e^{-iE_i t}$. Since the three masses are different, the three phases advance at **different rates**. The mixture that was a pure $\nu_e$ at the source therefore turns into a different mixture later, and that new mixture contains some $\nu_\mu$ and $\nu_\tau$.

## 4. The "three clocks" analogy

Picture each mass state $\nu_1,\nu_2,\nu_3$ as a **wave**, or equivalently as a **clock hand**, that rotates at its own rate. The rate depends on the mass, so the three rates differ slightly.

1. **At the source:** all three clocks are set to zero together. This particular combination of the three (with the weights given by $U$) *is* a pure $\nu_e$.
2. **During the journey:** the clock hands drift apart, because they rotate at different speeds.
3. **At the detector:** we ask how much the current combination looks like $\nu_e$, like $\nu_\mu$, and like $\nu_\tau$. The answer depends on the *relative positions of the clock hands*.
4. **Later:** when the hands line up again in the original pattern, the neutrino is a pure $\nu_e$ again. That is why the flavor **oscillates** back and forth.

## 5. Two-flavor oscillation (The Main Project)

The simplest case has only two flavors, say $\nu_e$ and $\nu_\mu$, and two mass states, $\nu_1$ and $\nu_2$. The mixing matrix is a plain rotation by one **mixing angle** $\theta$:

$$\begin{pmatrix}\nu_e\\ \nu_\mu\end{pmatrix}=\begin{pmatrix}\cos\theta & \sin\theta\\ -\sin\theta & \cos\theta\end{pmatrix}\begin{pmatrix}\nu_1\\ \nu_2\end{pmatrix}$$

**Step 1: start as $\nu_e$.**
$$|\nu(0)\rangle=|\nu_e\rangle=\cos\theta\,|\nu_1\rangle+\sin\theta\,|\nu_2\rangle$$

**Step 2: let it evolve.** Each mass state is an energy eigenstate, so it only picks up a phase (units $\hbar=c=1$):
$$|\nu(t)\rangle=\cos\theta\,e^{-iE_1t}|\nu_1\rangle+\sin\theta\,e^{-iE_2t}|\nu_2\rangle$$

**Step 3: amplitude to find $\nu_\mu$.** Using $\langle\nu_\mu|=-\sin\theta\langle\nu_1|+\cos\theta\langle\nu_2|$:
$$A(\nu_e\to\nu_\mu)=\langle\nu_\mu|\nu(t)\rangle=\sin\theta\cos\theta\left(e^{-iE_2t}-e^{-iE_1t}\right)$$

**Step 4: probability.** Square the magnitude and use $|e^{-ia}-e^{-ib}|^2=4\sin^2\!\frac{a-b}{2}$ and $4\sin^2\theta\cos^2\theta=\sin^22\theta$:
$$P(\nu_e\to\nu_\mu)=|A|^2=\sin^22\theta\;\sin^2\!\left(\frac{(E_2-E_1)\,t}{2}\right)$$

**Step 5: express it with measurable quantities.** For a light neutrino of momentum $p\approx E$ we have $E_i\approx p+\dfrac{m_i^2}{2E}$, so $E_2-E_1\approx\dfrac{\Delta m^2}{2E}$ with $\Delta m^2=m_2^2-m_1^2$. The neutrino travels at almost the speed of light, so $t\approx L$:

$$\boxed{P(\nu_e\to\nu_\mu)=\sin^2 2\theta\;\sin^2\!\left(\frac{\Delta m^2 L}{4E}\right)}\qquad P(\nu_e\to\nu_e)=1-P(\nu_e\to\nu_\mu)$$

**Practical units** (this is the formula experimentalists use):

$$P=\sin^22\theta\,\sin^2\!\left(1.267\,\frac{\Delta m^2[\text{eV}^2]\;L[\text{km}]}{E[\text{GeV}]}\right)$$

**What the formula tells us**

- $\sin^22\theta$ is the **amplitude**: how *much* the flavor changes (set by mixing).
- $\Delta m^2 L/E$ is the **phase**: how *fast* it changes with distance (set by the mass difference).
- If $\Delta m^2=0$ or $\theta=0$, nothing happens, so oscillation requires both mixing and different masses.




 # Neutrino oscillations with Qiskit
 **Idea:** a neutrino is created in a flavor state (here nu_e) but travels as a superposition of two mass states. Their phases drift apart with distance, so the flavor changes: P($\nu_e$ -> $\nu_{\mu}$) = $\sin^2(2θ)$ * $\sin^2(1.267 \frac{Δm² L }{ E})$.
 
**Qubit mapping:** |0> = $\nu_e$, |1> = $\nu_{\mu}$. 

**Circuit**: Ry(2θ) -> Rz(φ) -> Ry(-2θ) -> measure, with φ = Δm² L / (2E).

```python
##  Imports
import numpy as np
import matplotlib.pyplot as plt
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
```

```python
##  Parameters
theta = np.pi / 8       # mixing angle (22.5 degrees)
dm2 = 2.5*(10**(-3))    # Δm² in eV^2 (atmospheric value)
shots = 5000            # measurements per circuit
K = 1.267               # unit factor

backend = AerSimulator()
```

```python
##   Making a neutrino 
def make_circuit(theta, phi):
    ##One-qubit circuit for two-flavor oscillation.
    qc = QuantumCircuit(1, 1)
    qc.ry(2 * theta, 0)      # flavor state -> mass-state amplitudes (cosθ, sinθ)
    qc.rz(phi, 0)            # relative phase between the two mass states
    qc.ry(-2 * theta, 0)     # back to the flavor basis (exact inverse of first Ry)
    qc.measure(0, 0)
    return qc


def phase(dm2, LE):
    """Rz angle. The probability depends on sin^2(phi/2) = sin^2(K*dm2*L/E)."""
    return 2 * K * dm2 * LE
```

```python
##   Running the circuit with different relative phases
def run_point(theta, LE, shots=shots):
    qc = make_circuit(theta, phase(dm2, LE))
    counts = backend.run(qc, shots=shots).result().get_counts()
    return counts.get("1", 0) / shots
```

```python
##   Theoretical value of probablity of converting
def theory(theta, LE):
    return np.sin(2 * theta) ** 2 * np.sin(K * dm2 * LE) ** 2
```

```python
## Displaying the circuit
print(make_circuit(theta, phase(dm2, 500)).draw())
```

```python
## Generating relative phases and counting the probablity for each phase
LE_points = np.linspace(0, 2000, 50)          # L/E in km/GeV
P_sim = np.array([run_point(theta, x) for x in LE_points])
```

```python
## Theortical values of probablity for different relative phases
LE_fine = np.linspace(0, 2000, 1000)          # Extra values are taken to make a smoother curve
P_th = theory(theta, LE_fine)
```

```python
## Plotting the curves
plt.plot(LE_fine, P_th, color = 'black', label="theory")
plt.scatter(LE_points, P_sim, color="deeppink",
             label=f"Qiskit (Aer, {shots} shots)")
plt.xlabel("L/E (km/GeV)")
plt.ylabel(r"$P(\nu_e \to \nu_\mu)$")
plt.title(f"Two-flavor neutrino oscillation, θ = {np.degrees(theta):.1f}°")
plt.legend()
plt.show()
```

```python
## Checking the validity of program
# 1) L/E = 0: no distance travelled, so no conversion
print("L/E = 0 gives P = 0:",
      "OK" if run_point(theta, 0) == 0 else "FAILED")
 
# 2) theta = 0: no mixing, so the flavor never changes
print("theta = 0 gives P = 0 everywhere:",
      "OK" if all(run_point(0.0, x) == 0 for x in LE_points) else "FAILED")
 
# 3) theta = pi/4 (maximal mixing): P reaches ~1 at the first maximum
LE_max = np.pi / (2 * K * dm2)                 # where sin^2(K*dm2*L/E) = 1
p_max = run_point(np.pi / 4, LE_max)
print(f"theta = pi/4 reaches P = {p_max:.3f} at first maximum (expect ~1):","OK" if p_max > 0.97 else "FAILED")
dev = np.abs(P_sim - theory(theta, LE_points))
print(f"max deviation from theory = {dev.max():.4f}")
```
